**IN LOCAL**

4. Unzip the .zip file. You should then have 12 csoi matfiles. These files hold all the info for the waves you need. The scenarios seem to be the noisy, raw data whereas this is the cleaned, denoised ones. Below is how to extract the data for the first file and turn it into ML trainable data

```python
from scipy.io import loadmat
import numpy as np

data = loadmat('csoi/csoi_0_matfile.mat') # replace with the path to ur file
```

```python
iq = data['csoi_ComplexVoltageData'].flatten()  # (2.5M,)
print(f"Original Data Shape:        {iq.shape}")
print(f"Original Sample Data:       {iq[1]}")

# Split I and Q → shape (2.5M, 2)
I = np.real(iq)[:, np.newaxis]  
Q = np.imag(iq)[:, np.newaxis]
segments_long = np.concatenate([I, Q], axis=1)  # (2.5M, 2)

# Chop into 128-sample segments
segment_length = 128
X = []
for i in range(0, len(segments_long) - segment_length, 64):  # 50% overlap cause apparently that's common
    X.append(segments_long[i:i+segment_length])

X = np.array(X)

print(f"\nNew Data Shape:             {X.shape}")
print(f"New Sample Data Shape:      {X[1].shape}")
```
